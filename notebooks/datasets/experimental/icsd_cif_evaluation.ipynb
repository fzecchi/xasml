```python
"""Evaluation of the information read from ICSD CIF files."""

from pymatgen.io.cif import CifParser
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

from xasml import resource_path
```

```python
CIF_PATH = resource_path(
    "xasml:datasets/experimental/materials/FeO/ICSD_EntryWithCollCode76639.cif"
)
cif = CifParser(CIF_PATH)
structure = cif.parse_structures(primitive=False)[0]
print(structure)
spa = SpacegroupAnalyzer(structure, symprec=0.01, angle_tolerance=5)
print(spa.get_space_group_symbol())
symmetry_dataset = spa.get_symmetry_dataset()
print(symmetry_dataset)
print(spa.get_symmetrized_structure().equivalent_sites)
```
